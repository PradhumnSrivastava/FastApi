### **HTTP Stauts Codes**

# HTTP Status Codes

> HTTP status codes are **3-digit numbers** returned by a server to indicate the result of a client's request.

---

#### 1xx – Informational Responses

```text
Purpose:
Request has been received and the process is continuing.
```

| Status Code             | Meaning                                    |
| ----------------------- | ------------------------------------------ |
| 100 Continue            | Client should continue sending the request |
| 101 Switching Protocols | Server is switching protocols              |
| 102 Processing          | Server is processing the request           |

---

#### 2xx – Success Responses

```text
Purpose:
The request was successfully received and processed.
```

| Status Code    | Meaning                           |
| -------------- | --------------------------------- |
| 200 OK         | Request succeeded                 |
| 201 Created    | New resource created successfully |
| 202 Accepted   | Request accepted for processing   |
| 204 No Content | Success, but nothing to return    |

---

#### 3xx – Redirection Responses

```text
Purpose:
Additional action is required to complete the request.
```

| Status Code            | Meaning                                   |
| ---------------------- | ----------------------------------------- |
| 301 Moved Permanently  | Resource permanently moved                |
| 302 Found              | Temporary redirect                        |
| 304 Not Modified       | Cached resource can be used               |
| 307 Temporary Redirect | Temporary redirect preserving HTTP method |

---

#### 4xx – Client Error Responses

```text
Purpose:
The client sent an invalid or incorrect request.
```

| Status Code              | Meaning                                |
| ------------------------ | -------------------------------------- |
| 400 Bad Request          | Invalid request syntax                 |
| 401 Unauthorized         | Authentication required                |
| 403 Forbidden            | Access denied                          |
| 404 Not Found            | Resource does not exist                |
| 405 Method Not Allowed   | HTTP method not allowed                |
| 409 Conflict             | Conflict with current resource state   |
| 422 Unprocessable Entity | Validation error (commonly in FastAPI) |

---

#### 5xx – Server Error Responses

```text
Purpose:
The server failed to process a valid request.
```

| Status Code               | Meaning                               |
| ------------------------- | ------------------------------------- |
| 500 Internal Server Error | Generic server-side error             |
| 501 Not Implemented       | Feature not supported                 |
| 502 Bad Gateway           | Invalid response from upstream server |
| 503 Service Unavailable   | Server temporarily unavailable        |
| 504 Gateway Timeout       | Upstream server timeout               |

---

#### Quick Summary

```text
1xx → Information
2xx → Success
3xx → Redirection
4xx → Client Error
5xx → Server Error
```
