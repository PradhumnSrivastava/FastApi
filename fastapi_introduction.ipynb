## **FastApi** **Application Programming Interface**


APIs are mechnaism that enable two software components- such as the frontend and backend of an application- to communicate with each other using a defined set of rules, protocols and data formats.

### **Bridge between AI model and real world applications**

### **Core Technical ObjectivesMaximize**

Speed and Performance: Aiming to be one of the fastest Python frameworks available. It handles asynchronous concurrency natively via ASGI, making it optimal for heavy I/O-bound processes and machine learning deployments.

Increase Development Velocity: Designed to boost overall feature development speed by an estimated 200% to 300%. It cuts down boilerplate code by generating multiple features from single parameter declarations.

Drastically Reduce Code Bugs: Seeking to eliminate roughly 40% of developer-induced errors. It enforces data type validation at runtime using Python type hints, ensuring data compatibility before processing.

Automate Documentation: Eliminating manual API documenting tasks. It natively generates interactive, web-based UI playgrounds—like Swagger UI and ReDoc—straight from the code structure.

Enforce Open Standards: Building apps completely compatible with open-source API requirements. The environment relies entirely on the OpenAPI specification and JSON Schema standards.Architecture & Usability Objective

### **Monolithic Architecture**

Monolithic architecture is a software design approach where all components of an application — such as user interface, business logic, database handling, and APIs — are built together as a single unified system. In this architecture, the entire application runs as one codebase and is deployed as one unit. If one small part needs modification, the whole application often needs to be rebuilt and redeployed. Monolithic systems are simpler to develop initially and are suitable for small projects, but as applications grow, they become harder to scale, maintain, and update. Many traditional applications were built using monolithic architecture before microservices became popular.


### **FastAPI: Core Libraries and Their Roles**

FastAPI is not built from scratch. It is mainly built on top of several powerful Python libraries that work together to provide high performance, data validation, API development, and asynchronous processing.

#### **1. Starlette**
What is Starlette?

Starlette is a lightweight ASGI (Asynchronous Server Gateway Interface) framework.

Role in FastAPI

FastAPI uses Starlette for:

- Routing (/users, /products, etc.)
- Middleware
- Request and Response handling
- Background tasks
- WebSockets
- Session management
- Event handling



Example

When you write:

```python
@app.get("/hello")
def hello():
    return {"message": "Hello"}
```

The actual routing and HTTP request handling are performed by Starlette.

#### **2. Pydantic**
What is Pydantic?

Pydantic is a data validation and serialization library.

Role in FastAPI

FastAPI uses Pydantic for:

- Request validation
- Response validation
- Type checking
- Automatic data conversion
- API documentation generation

Example

```python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int
```

If a user sends:

```python
{
    "name": "John",
    "age": "25"
}
```

Pydantic automatically converts "25" into 25.

#### **3. ASGI**
What is ASGI?

ASGI (Asynchronous Server Gateway Interface) is the modern successor of WSGI.

Role in FastAPI

ASGI allows:

- Async programming
- Concurrent requests
- WebSockets
- High-performance APIs
